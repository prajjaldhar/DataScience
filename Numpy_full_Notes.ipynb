# Capstone: Mini Netflix Churn + Agentic AI Feature Pipeline

Use NumPy to build a compact feature pipeline:
- customer age
- salary
- watch hours
- subscription price
- support tickets
- churn label
- token IDs
- agent confidence

The exercise intentionally combines **creation → filtering → broadcasting → statistics → sorting → vector operations**.

```python
n = 20
age = np.random.randint(18, 61, n)
salary = np.random.randint(25000, 120001, n)
watch_hours = np.random.uniform(0, 50, n)
plan_price = np.random.choice(np.array([149, 199, 499, 649]), n)
support_tickets = np.random.randint(0, 8, n)
churn = np.random.rand(n)

features = np.column_stack([
    age,
    salary,
    watch_hours,
    plan_price,
    support_tickets
])

print("Feature matrix shape:", features.shape)
print("First 5 rows:")
print(features[:5])
```

```python
# Example churn-risk score using broadcasting
risk = (
    (watch_hours < 8) * 0.35 +
    (support_tickets >= 4) * 0.25 +
    (plan_price == 149) * 0.10 +
    (age > 45) * 0.10 +
    (salary < 40000) * 0.05
)

risk = np.clip(risk, 0, 1)

top_idx = np.argsort(risk)[-5:][::-1]

print("Top 5 risk customers:", top_idx)
print("Risk scores:", risk[top_idx])
print("Actual churn labels:", churn[top_idx])
```

```python
# Token / embedding style example
token_ids = np.array([101, 2057, 2024, 1037, 2742, 102])
embedding_table = np.random.normal(0, 1, (3000, 8))

selected_embeddings = embedding_table[token_ids]

print("Token IDs:", token_ids)
print("Embedding shape:", selected_embeddings.shape)

# Mean pooling — a common simple NLP operation
sentence_embedding = selected_embeddings.mean(axis=0)
print("Sentence embedding:", sentence_embedding)
```

```python
# Agentic AI-style decision example
agent_confidence = np.random.uniform(0, 1, 20)
tool_cost = np.random.uniform(0, 100, 20)

decision_score = agent_confidence - 0.003 * tool_cost

best_agent_step = np.argmax(decision_score)

print("Best step:", best_agent_step)
print("Confidence:", agent_confidence[best_agent_step])
print("Tool cost:", tool_cost[best_agent_step])
print("Decision score:", decision_score[best_agent_step])
```
