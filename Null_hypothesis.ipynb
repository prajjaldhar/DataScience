```python
import pandas as pd
import numpy as np
from scipy.stats import pearsonr
```

## Jo Zada padhta hai uske marks bhi zada ata hai.

```python
hours = [10,8,4,5,2,8,10,12,9]
marks = [45,52,68,75,88,78,56,48,66]
student={
    "hours":hours,
    "marks":marks
}
marks = pd.DataFrame(student)
marks
```

```python
marks.describe()
```

```python
marks.mean()
```

## H0 : Hours and marks ke bich relationship exist karti nahi hai
## H1 : karti hai

```python
r,p_val=(pearsonr(marks["hours"],marks["marks"]))
print("Correlation: ",r)
print("p-value",p_val)
```

```python
alpha = 0.05
if p_val<alpha:
  print("Reject H0")
else :
  print("Fail to reject H0")
```
