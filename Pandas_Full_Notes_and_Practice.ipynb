## 1. What is Pandas?

**Pandas** is a Python library for data manipulation and analysis.

Its two fundamental structures are:

- **Series** → one-dimensional labeled data
- **DataFrame** → two-dimensional labeled/tabular data

Typical data-analysis workflow:

**Load → Inspect → Clean → Transform → Analyze → Export**

## 3. Series Attributes

Important attributes: `index`, `values`, `dtype`, `shape`, `size`, `ndim`, `name`, `empty`, `hasnans`. `unique()` and `nunique()` are methods.

```python
sales = pd.Series([1200, 850, 910, 1500,850],
                  index=["Mon", "Tue", "Wed", "Thu","Fri"],
                  name="daily_sales")

print("index:", sales.index)
print("values:", sales.values)
print("dtype:", sales.dtype)
print("shape:", sales.shape)
print("size:", sales.size)
print("ndim:", sales.ndim)
print("name:", sales.name)
print("empty:", sales.empty)
print("hasnans:", sales.hasnans)
print("unique:", sales.unique())
print("is unique:", sales.is_unique)
print("nunique:", sales.nunique())
```

## 5. `loc` vs `iloc`

`loc` is label-based. `iloc` is position-based. `loc` label slices are generally inclusive; `iloc` follows Python's exclusive stop rule.

## 19. Merge

```python
customers = pd.DataFrame({
    "customer_id":[1,2,3],
    "name":["Aman","Riya","Kabir"]
})
orders = pd.DataFrame({
    "customer_id":[1,1,2,4],
    "amount":[500,800,1200,700]
})
```

### Merge scenario 1

```python
pd.merge(customers, orders, on="customer_id")
```

### Merge scenario 2

```python
pd.merge(customers, orders, on="customer_id", how="left")
```

### Merge scenario 3

```python
pd.merge(customers, orders, on="customer_id", how="right")
```

### Merge scenario 4

```python
pd.merge(customers, orders, on="customer_id", how="outer")
```

### Merge scenario 5

```python
pd.merge(customers, orders, on="customer_id", how="inner")
```

### Merge scenario 6

```python
pd.merge(customers, orders, left_on="customer_id", right_on="customer_id")
```

### Merge scenario 7

```python
pd.merge(pd.DataFrame({"id":[1,2],"value":[10,20]}), pd.DataFrame({"id":[1,2],"value":[100,200]}), on="id", suffixes=("_left","_right"))
```

### Merge scenario 8

```python
customers.set_index("customer_id").join(orders.set_index("customer_id"), how="left")
```

### Merge scenario 9

```python
pd.merge(customers, orders, on="customer_id", indicator=True, how="outer")
```

### Merge scenario 10

```python
pd.merge(customers, orders, on="customer_id", validate="one_to_many")
```
