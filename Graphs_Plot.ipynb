## 2. Box Plot

- Box Plot ka main purpose hota hai → data ki distribution, median, spread aur outliers dekhna.

Sales dataset mein example:

- Sales ki distribution aur outliers check karna.

```python
plt.boxplot(df["Sales"])

plt.ylabel("Sales")
plt.title("Distribution of Sales")

plt.show()
```

        •       ← Outlier
        |
    ┌───────┐
    │       │
    │   ─   │ ← Median
    │       │
    └───────┘
        |
        |

| **Part**          | **Meaning**        |
| ----------------- | ------------------ |
| **Box**           | Q1 → Q3            |
| **Middle Line**   | Median             |
| **Lower Whisker** | Lower range        |
| **Upper Whisker** | Upper range        |
| **Dots**          | Potential outliers |


Important terms

- Q1 → 25th percentile
- Median (Q2) → 50th percentile
- Q3 → 75th percentile

IQR = Q3 - Q1

Potential outliers generally:

- X < Q1 - 1.5 × IQR
- X > Q3 + 1.5 × IQR

```python
sns.boxplot(data=df, y="Sales")

plt.ylabel("Sales")
plt.title("Distribution of Sales")

plt.show()
```

## 8. Area Plot

Area Plot ka main purpose hota hai → trend ke saath value ka magnitude visually show karna.

Line Plot jaisa hi hai, lekin line ke neeche ka area fill ho jata hai.

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

monthly_sales = df.resample(
    "ME",
    on="Order_Date"
)["Sales"].sum()
```

```python
plt.fill_between(
    monthly_sales.index,
    monthly_sales.values
)

plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.title("Monthly Sales Trend")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.stackplot(
    monthly_sales.index,
    monthly_sales.values
)

plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.title("Monthly Sales Area Plot")
plt.xticks(rotation=45)

plt.show()
```

Seaborn mein dedicated area plot function nahi hai.

Area plot ke liye generally Matplotlib/Pandas use karte hain.

```python
monthly_sales.plot.area()

plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.title("Monthly Sales Area Plot")

plt.show()
```

## 9. Violin Plot

- Violin Plot ka main purpose hota hai → data ki distribution + density ko visualize karna.

- Box Plot jaisa hi hai, lekin isme data kis range mein kitna concentrated hai ye bhi clearly dikhta hai.

```python
sns.violinplot(
    data=df,
    y="Sales"
)

plt.ylabel("Sales")
plt.title("Sales Distribution")

plt.show()
```

Product-wise Sales Distribution

Violin ki Shape ka Meaning

Violin jitna wide hai → us range par zyada data points/density.

Violin jitna narrow hai → us range par kam data/density

## Box Plot vs Violin Plot

- Box Plot	Median, quartiles, spread, outliers
- Violin Plot	Distribution + density + spread

Violin Plot is used to visualize the distribution and density of numerical data, often across different categories.

```python
sns.violinplot(
    data=df,
    x="Product",
    y="Sales"
)

plt.xticks(rotation=45)
plt.xlabel("Product")
plt.ylabel("Sales")
plt.title("Sales Distribution by Product")

plt.show()
```

## 10. Heatmap

- Heatmap ka main purpose hota hai → multiple numerical variables ke beech relationship/correlation ko color ke through visualize karna.

- Sales dataset mein hum correlation matrix ko heatmap mein dikha sakte hain.

Correlation Matrix

Pehle numerical columns select karte hain:

```python
corr = df[
    ["Units_Sold", "Unit_Price", "Sales", "Profit", "Discount", "Customer_Rating"]
].corr()

print(corr)
```

Ye har pair ke beech correlation value dega.

```python
corr
```

annot=True kya karta hai?

Cells ke andar correlation values show karta hai.

Example:
| **Variable** | **Sales** | **Profit** |
| ------------ | --------: | ---------: |
| **Sales**    |      1.00 |       0.82 |
| **Profit**   |      0.82 |       1.00 |


```python
sns.heatmap(
    corr,
    annot=True
)

plt.title("Correlation Heatmap")
plt.show()
```

fmt=".2f" → values ko 2 decimal places tak show karega.
fmt matlab formatting

Correlation ki value generally -1 se +1 ke beech hoti hai.

Correlation	Meaning
- +1	Perfect positive relationship
- 0	No linear relationship
- -1	Perfect negative relationship
- +0.8	Strong positive relationship
- -0.8	Strong negative relationship

Units_Sold ↑  →  Sales ↑

To correlation positive ho sakta hai.

| **Relationship**             | **Correlation** | **Insight**                       |
| ---------------------------- | --------------: | --------------------------------- |
| **Sales ↔ Profit**           |        **0.94** | Very strong positive relationship |
| **Unit_Price ↔ Sales**       |        **0.82** | Strong positive relationship      |
| **Unit_Price ↔ Profit**      |        **0.75** | Strong positive relationship      |
| **Units_Sold ↔ Sales**       |        **0.40** | Moderate positive relationship    |
| **Units_Sold ↔ Profit**      |        **0.38** | Moderate positive relationship    |
| **Discount ↔ Profit**        |       **-0.10** | Very weak negative relationship   |
| **Discount ↔ Sales**         |      **-0.055** | Almost no linear relationship     |
| **Customer_Rating ↔ Sales**  |       **-0.10** | Very weak negative relationship   |
| **Customer_Rating ↔ Profit** |       **-0.10** | Very weak negative relationship   |


iska matlab ye nahi ki Sales increase karna automatically Profit increase karta hai. Correlation sirf variables ke linear association ko measure karta hai; cause-and-effect prove nahi karta.

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

## 12. Joint Plot

Joint Plot ka main purpose hota hai → 2 numerical variables ke beech relationship + dono variables ki individual distribution ek saath dekhna.

Matlab Scatter Plot + Distribution Plot ek hi visualization mein.

```python
sns.jointplot(
    data=df,
    x="Units_Sold",
    y="Sales"
)

plt.show()
```

- Center → Scatter Plot
- Top → Units_Sold ki distribution
- Right → Sales ki distribution

kind="reg" regression line add karta hai.

```python
sns.jointplot(
    data=df,
    x="Units_Sold",
    y="Sales",
    kind="reg"
)

plt.show()
```

Different Joint Plot Types

Seaborn mein kind change karke visualization change kar sakte hain:

kind	Visualization
- "scatter"	Scatter Plot
- "reg"	Scatter + Regression Line
- "hist"	Histogram
- "kde"	KDE / Density
- "hex"	Hexagonal bins

```python
sns.jointplot(
    data=df,
    x="Unit_Price",
    y="Sales",
    kind="reg"
)

plt.show()
```

```python
sns.jointplot(
    data=df,
    x="Unit_Price",
    y="Sales",
    kind="hist"
)

plt.show()
```

```python
sns.jointplot(
    data=df,
    x="Profit",
    y="Sales",
    kind="reg"
)

plt.show()
```
