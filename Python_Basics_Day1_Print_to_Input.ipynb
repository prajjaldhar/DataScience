## 1. Print Statement

```python
print("Hello World")
```

## 2. Printing Numbers

```python
print(10)
print(20)
print(10+20)
```

## 3. Multiple Values

```python
print("Prajjal",36,"India")
```

## 4. Escape Characters

```python
print("Hello\nWorld")
print("Python\tProgramming")
print("He said \"Hello\"")
```

## 5. Variables

```python
name="Rahul"
age=22
print(name)
print(age)
```

## 6. Data Types

```python
name="Rahul"
age=22
height=5.8
is_student=True

print(type(name))
print(type(age))
print(type(height))
print(type(is_student))
```

## 7. Multiple Assignment

```python
a,b,c=10,20,30
print(a,b,c)
```

## 8. Swapping

```python
a=10
b=20
a,b=b,a
print(a,b)
```

## 9. String Concatenation

```python
first="Hello"
second="World"
print(first+" "+second)
```

## 10. String Repetition

```python
print("*"*10)
```

## 11. Taking Input

```python
name=input("Enter your name: ")
print(name)
```

## 12. Integer Input

```python
age=int(input("Enter age: "))
print(age)
print(type(age))
```

## 13. Float Input

```python
salary=float(input("Enter salary: "))
print(salary)
```

## 14. Two Numbers

```python
a=int(input("Enter first number: "))
b=int(input("Enter second number: "))
print("Sum =",a+b)
```

## 15. Three Numbers One Line

```python
a,b,c=map(int,input("Enter three numbers: ").split())
print(a,b,c)
```

## 16. List Input

```python
numbers=list(map(int,input("Enter numbers: ").split()))
print(numbers)
```
