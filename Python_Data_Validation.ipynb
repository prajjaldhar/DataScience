## 1. `==` kya check karta hai?

`==` mainly **values compare** karta hai.

```python
x == 10
```

iska matlab hai: kya `x` ki value `10` ke equal hai?

````python
x = 10

print(x == 10)
print(x == 20)
print("10" == 10)
````

### Important

`==` khud se type validation nahi karta.

```python
"10" == 10
```

`False` hoga kyunki values/types compatible nahi hain.

## 2. `type(x) == int`

`type()` se hum variable ka type check kar sakte hain.

```python
type(x) == int
```

Yeh check karta hai ki `x` ka exact type `int` hai ya nahi.

````python
x = 10

print(type(x) == int)
print(type(x) == float)
print(type(x) == str)
````

````python
def square(num):
    if type(num) == int:
        return num * num
    else:
        print("Only integer is allowed")


print(square(5))
print(square("5"))
print(square(5.5))
````

## 3. `type(x) is int`

`is` identity comparison ke liye use hota hai.

Exact type validation mein:

```python
type(x) is int
```

use kar sakte hain.

````python
x = 10

print(type(x) is int)
print(type(x) is float)
````

### `type(x) == int` vs `type(x) is int`

Normal exact-type checking mein dono ka result practically same hoga:

```python
type(x) == int
type(x) is int
```

Beginner code mein `type(x) == int` easy to understand hai. Python mein exact type identity check ke liye `type(x) is int` bhi common hai.

## 4. `isinstance()`

`isinstance()` type checking ka powerful aur recommended way hai.

```python
isinstance(value, type)
```

````python
x = 10

print(isinstance(x, int))
print(isinstance(x, float))
print(isinstance(x, str))
````

````python
def square(num):
    if isinstance(num, int):
        return num * num

    raise TypeError("num must be an integer")


print(square(5))
````

## 5. `type()` aur `isinstance()` ka important difference

`bool`, Python mein `int` ka subclass hai.

Isliye:

````python
x = True

print(type(x) == int)
print(type(x) is int)
print(isinstance(x, int))
````

Output:

```text
False
False
True
```

Isliye:

- `type(x) == int` → exact type
- `type(x) is int` → exact type
- `isinstance(x, int)` → inheritance/subclass ko bhi allow karta hai

## 6. Multiple types allow karna

Agar function ko `int` **ya** `float` dono accept karne hain, `isinstance()` ke saath tuple use kar sakte hain.

````python
def calculate_price(price):

    if not isinstance(price, (int, float)):
        raise TypeError("price must be int or float")

    return price * 1.18


print(calculate_price(100))
print(calculate_price(100.5))
````

````python
# Invalid examples

# calculate_price("100")
# calculate_price([100])
````

## 7. Sirf `list` accept karna

Function ke input ko validate kar sakte hain:

````python
def calculate_total(numbers):

    if type(numbers) is not list:
        raise TypeError("numbers must be a list")

    return sum(numbers)


print(calculate_total([10, 20, 30]))
````

## 8. List ke andar ke elements validate karna

Sirf yeh check karna enough nahi hai ki input list hai.

Humein list ke andar ka data bhi validate karna pad sakta hai.

````python
def calculate_total(numbers):

    if type(numbers) is not list:
        raise TypeError("numbers must be a list")

    for num in numbers:
        if type(num) is not int:
            raise TypeError("Every element must be an integer")

    return sum(numbers)


print(calculate_total([10, 20, 30]))
````

````python
# Invalid examples:

# calculate_total([10, 20, "30"])
# calculate_total([10, 20, 30.5])
# calculate_total((10, 20, 30))
````

## 9. Type validation + value validation

Data validation ke do important parts hain:

1. **Type validation** → data kis type ka hai?
2. **Value validation** → data ki value allowed range mein hai ya nahi?

````python
def process_marks(marks):

    if type(marks) is not list:
        raise TypeError("marks must be a list")

    for mark in marks:

        if type(mark) is not int:
            raise TypeError("Every mark must be an integer")

        if mark < 0 or mark > 100:
            raise ValueError("Marks must be between 0 and 100")

    return sum(marks) / len(marks)


print(process_marks([70, 80, 90]))
````

## 10. `TypeError` vs `ValueError`

### TypeError
Jab **data ka type galat** ho.

Example:

```python
process_marks([70, "80", 90])
```

### ValueError
Jab type sahi ho, lekin **value invalid** ho.

Example:

```python
process_marks([70, 150, 90])
```

`150` integer hai, lekin marks ki valid range `0–100` hai.

````python
def check_age(age):

    if type(age) is not int:
        raise TypeError("Age must be an integer")

    if age < 0 or age > 120:
        raise ValueError("Age must be between 0 and 120")

    return "Valid age"


print(check_age(25))
````

## 11. String validation

Function ko sirf string accept karwani ho:

````python
def greet(name):

    if type(name) is not str:
        raise TypeError("name must be a string")

    return "Hello " + name


print(greet("Prajjal"))
````

## 12. Empty string validation

Type sahi hone ke baad value bhi validate kar sakte hain.

````python
def greet(name):

    if type(name) is not str:
        raise TypeError("name must be a string")

    if name == "":
        raise ValueError("name cannot be empty")

    return "Hello " + name


print(greet("Prajjal"))
````

## 13. Boolean validation

Agar function ko strictly `True` ya `False` chahiye:

````python
def check_status(status):

    if type(status) is not bool:
        raise TypeError("status must be True or False")

    return "Status accepted"


print(check_status(True))
print(check_status(False))
````

## 14. Practical example: Student data validation

````python
def calculate_average(marks):

    if type(marks) is not list:
        raise TypeError("marks must be a list")

    if len(marks) == 0:
        raise ValueError("marks list cannot be empty")

    for mark in marks:

        if type(mark) is not int:
            raise TypeError("Every mark must be an integer")

        if mark < 0 or mark > 100:
            raise ValueError("Every mark must be between 0 and 100")

    return sum(marks) / len(marks)


print(calculate_average([80, 75, 90, 85]))
````

## 15. Validation flow

Function mein validation ko is order mein sochna useful hai:

```text
Function Input
      ↓
Container / Main Type Check
      ↓
Element Type Check
      ↓
Value / Range Check
      ↓
Actual Function Logic
      ↓
Return Result
```

# Quick Revision 🧠

| Code | Meaning |
|---|---|
| `x == 10` | Value comparison |
| `type(x) == int` | Exact type check |
| `type(x) is int` | Exact type check |
| `isinstance(x, int)` | Type/subclass check |
| `isinstance(x, (int, float))` | Multiple allowed types |
| `raise TypeError()` | Wrong data type |
| `raise ValueError()` | Correct type but invalid value |

```python

    raise TypeError()

    raise ValueError()
```
