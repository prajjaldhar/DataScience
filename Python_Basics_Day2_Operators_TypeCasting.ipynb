# 🐍 Python Basics - Day 2
## Operators & Type Casting

Learn arithmetic, comparison, logical operators, assignment operators, membership, identity and type casting.

## 1. Arithmetic Operators

```
a=10
b=3
print("Addition:",a+b)
print("Subtraction:",a-b)
print("Multiplication:",a*b)
print("Division:",a/b)
print("Floor Division:",a//b)
print("Modulus:",a%b)
print("Power:",a**b)
```

## 2. Comparison Operators

```
a=10
b=20
print(a==b)
print(a!=b)
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
```

## 3. Logical Operators

```
x=True
y=False
print(x and y)
print(x or y)
print(not x)
```

## 4. Assignment Operators

```
x=10
x+=5
print(x)
x*=2
print(x)
x-=4
print(x)
```

## 5. Membership Operators

```
name="Python Programming"
print("Python" in name)
print("Java" not in name)
```

## 6. Identity Operators

```
a=[1,2]
b=a
c=[1,2]
print(a is b)
print(a is c)
print(a==c)
```

## 7. Type Casting

```
num="100"
print(type(num))
num=int(num)
print(type(num))
print(num+50)
```

## 8. int(), float(), str()

```
a=10
b=float(a)
c=str(a)
print(b,type(b))
print(c,type(c))
```

## 9. Mini Program

```
a=int(input("Enter first number: "))
b=int(input("Enter second number: "))
print("Sum:",a+b)
print("Difference:",a-b)
print("Product:",a*b)
print("Division:",a/b)
```
