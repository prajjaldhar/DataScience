# 📱 WhatsApp Business Analytics:

### problem Statement:
A company uses WhatsApp for customer support and sales. Management wants to understand response speed, resolution efficiency, customer satisfaction, sales value, and agent performance.

**constraint:** NumPy only. No Pandas.

### NumPy concepts covered
- `np.arange`
- `np.random`
- `np.concatenate`
- `np.stack`
- Broadcasting
- Slicing
- Boolean indexing
- Aggregations and basic statistics

All business variables below are numerical.


````python
import numpy as np
print("NumPy version:", np.__version__)
````

## 1. Create the Business Dataset

We will create **20 WhatsApp business chats** and **12 numerical columns**.

| Column | Meaning |
|---|---|
| Chat_ID | Chat identifier |
| Response_Time | Minutes until first response |
| Resolution_Time | Minutes until issue resolution |
| Messages | Messages in conversation |
| Customer_Wait | Minutes customer waited |
| Satisfaction | Rating from 1–5 |
| Order_Value | Order value in ₹ |
| Discount | Discount in ₹ |
| Agent_Efficiency | Efficiency score |
| Followups | Number of follow-ups |
| Escalations | Number of escalations |
| Conversation_Hours | Conversation duration in hours |

We will keep the actual analysis matrix purely numerical.


````python
# 20 chat IDs
chat_id = np.arange(1, 21)

# Generate numerical business data using np.random
response_time = np.random.randint(2, 31, 20)
resolution_time = np.random.randint(10, 181, 20)
messages = np.random.randint(5, 61, 20)
customer_wait = np.random.randint(1, 46, 20)
satisfaction = np.random.randint(1, 6, 20)
order_value = np.random.randint(500, 10001, 20)
discount = np.random.randint(0, 1501, 20)
agent_efficiency = np.random.randint(60, 101, 20)
followups = np.random.randint(0, 7, 20)
escalations = np.random.randint(0, 4, 20)
conversation_hours = np.random.randint(1, 25, 20)

print("Chat IDs:", chat_id)
print("Response times:", response_time)
````

## 2. `np.arange` — Create Chat IDs

A business system gives each conversation a sequential ID.

### Practice

## 3. `np.concatenate` — Combine Data

Imagine the company exported two batches of chats.

**Question:** How can we combine the two batches into one array?


````python
batch_1 = np.arange(1, 11)
batch_2 = np.arange(11, 21)

all_chats = np.concatenate((batch_1, batch_2))

print("Batch 1:", batch_1)
print("Batch 2:", batch_2)
print("Combined:", all_chats)
````

## 4. `np.stack` — Build a Business Matrix

Now imagine each metric is one row. We want to place all metrics into a 2D matrix.

### Think:
`concatenate` → join arrays along an existing axis.

`stack` → create a new axis and place arrays together.


````python
sample_response = response_time[:5]
sample_resolution = resolution_time[:5]
sample_messages = messages[:5]

business_stack = np.stack(
    (sample_response, sample_resolution, sample_messages)
)

print("Stacked shape:", business_stack.shape)
print(business_stack)
````

## 5. Build the Complete Numerical Dataset

Each **row = one WhatsApp conversation**.

Each **column = one business metric**.

This is our main NumPy dataset.


````python
data = np.stack((
    chat_id,
    response_time,
    resolution_time,
    messages,
    customer_wait,
    satisfaction,
    order_value,
    discount,
    agent_efficiency,
    followups,
    escalations,
    conversation_hours
), axis=1)

print("Shape:", data.shape)
# print(*["chat_id",
#     "resp_time",
#     "resolu_time",
#     "msg",
#     "cust_wait",
#     "satftn",
#     "ord_val",
#     "dis",
#     "agnt_eff",
#     "follups",
#     "escltns",
#     "convrsn_hrs"],sep="|")
print(data)
````

#  6. Slicing

Management asks for only the **first 5 conversations**.

Then they ask for:
- response time and resolution time of all chats
- the first 10 chats
- chats 11–15
- order value for the last 5 chats


````python
first_5 = data[:5]
response_resolution = data[:, 1:3]
first_10 = data[:10]
chats_11_to_15 = data[10:15]
last_5_order_value = data[-5:, 6]

print("First 5:\n", first_5)
print("\nResponse + Resolution:\n", response_resolution)
print("\nFirst 10:\n", first_10)
print("\nChats 11–15:\n", chats_11_to_15)
print("\nLast 5 order values:", last_5_order_value)
````

# 7. Boolean Indexing

A manager asks:

> **Which conversations took more than 20 minutes to get a first response?**


````python
slow_response = data[data[:, 1] > 20]

print("Chats with response time > 20 minutes:")
print(slow_response)
````

# 8. Broadcasting

Suppose the company introduces a new service rule:

> **Every conversation gets an automatic 5-minute service buffer added to the recorded customer wait time.**

Use broadcasting to apply the value to the entire array.


````python
wait_with_buffer = data[:, 4] + 5

print("Original wait:", data[:, 4])
print("After 5-minute buffer:", wait_with_buffer)
````

## Broadcasting with a Business Matrix

The management team wants to increase every agent's efficiency score by **3 points** for a training simulation.

No loop. No Pandas.


````python
simulated_efficiency = data[:, 8] + 3

print("Original efficiency:", data[:, 8])
print("Simulated efficiency:", simulated_efficiency)
````

# Final Business Challenge

You are the **Data Analyst for the company's WhatsApp operations team**.

Management asks:

> **"Are our WhatsApp conversations efficient, and where should we improve?"**

Using only NumPy, produce these 5 metrics:

1. Average response time
2. Average resolution time
3. Average satisfaction
4. Total order value
5. Percentage of chats with response time > 20 minutes

Then write **3 business insights** based on the numbers.

### Rule
Don't just give numbers.

Convert:

**Number → Observation → Business Decision**

Example:

> Response time is high → customers wait longer → increase support capacity during peak periods.


````python
# Final challenge workspace

avg_response = np.mean(data[:, 1])
avg_resolution = np.mean(data[:, 2])
avg_satisfaction = np.mean(data[:, 5])
total_order_value = np.sum(data[:, 6])

slow_percentage = np.mean(data[:, 1] > 20) * 100

print("Average response time:", avg_response)
print("Average resolution time:", avg_resolution)
print("Average satisfaction:", avg_satisfaction)
print("Total order value: ₹", total_order_value)
print("Chats with response > 20 min:", slow_percentage, "%")
````
