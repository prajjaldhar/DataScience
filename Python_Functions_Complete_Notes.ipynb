## 1. What is a Function?

A function is a reusable block of code that performs a specific task.

### Syntax
```python
def function_name():
    # code
```

````python
def say_hello():
    print("Hello")

say_hello()
say_hello()
say_hello()
````

## 2. Function With Parameters

A **parameter** receives data inside the function. The supplied value is called an **argument**.

````python
def greet(name):
    print("Hello", name)

greet("Prajjal")
greet("Rahul")
greet("Amit")
````

## 3. Multiple Parameters

````python
def add(a, b):
    print(a + b)

add(10, 20)
add(50, 30)
````

## 4. `return` — How Does a Function Decide What to Return?

A function returns whatever expression is written after `return`.

```python
def add(a, b):
    return a + b
```

The expression after `return` is evaluated and its result is sent back to the caller.

````python
def add(a, b):
    return a + b

result = add(10, 20)
print(result)
````

````python
def square(x):
    return x * x

def get_name():
    return "Prajjal"

def check_number(x):
    if x > 0:
        return "Positive"
    else:
        return "Negative"

print(square(5))
print(get_name())
print(check_number(10))
print(check_number(-3))
````

## 5. `print()` vs `return`

- `print()` → displays a value
- `return` → sends a value back
- If a function has no `return`, Python returns `None`

````python
def add_print(a, b):
    print(a + b)

result = add_print(10, 20)
print("result =", result)
````

````python
def add_return(a, b):
    return a + b

result = add_return(10, 20)
print("result =", result)
````

## 6. Immutable Variables — Changes Inside the Function Don't Come Back

Integers, floats and strings are immutable.

When we do `x = 100` inside a function, the local parameter is rebound. The outside variable is not changed.

````python
def change_number(x):
    x = 100

num = 10
change_number(num)

print(num)
````

### Conceptual picture

```text
Before:
num → 10

Inside function:
x → 10

x = 100

After:
num → 10
x → 100
```

This is often described as **value-like behavior**, but Python's technical model is object-reference passing by sharing.

````python
def change_price(price):
    price = 99.99

price = 50.5
change_price(price)

print(price)
````

````python
def change_name(name):
    name = "Rahul"

name = "Prajjal"
change_name(name)

print(name)
````

## 7. Mutable / Non-Primitive Objects

Lists are mutable. A function can modify the existing list object, so the outside variable can see the modification.

````python
def add_number(numbers):
    numbers.append(100)

my_list = [10, 20, 30]
add_number(my_list)

print(my_list)
````

### Conceptual picture

```text
my_list ───────┐
               ↓
          [10, 20, 30]
               ↑
numbers ───────┘

numbers.append(100)
```

Both names refer to the same list object.

````python
def update_marks(marks):
    marks[0] = 100

marks = [50, 60, 70]
update_marks(marks)

print(marks)
````

## 8. Python's Actual Model: Object Reference / Call by Sharing

Do **not** treat this as a strict Python rule:

```text
int  = call by value
list = call by reference
```

The more accurate explanation is:

> Python passes references to objects by sharing.

| Object | Mutable? | Can a function modify the existing object? |
|---|---:|---:|
| `int` | ❌ | ❌ |
| `float` | ❌ | ❌ |
| `str` | ❌ | ❌ |
| `tuple` | ❌ | ❌ |
| `list` | ✅ | ✅ |
| `dict` | ✅ | ✅ |
| `set` | ✅ | ✅ |

## 9. Swap Two Numbers — First Attempt

Because integers are immutable, changing local parameters does not change the outside variables.

````python
def swap(a, b):
    temp = a
    a = b
    b = temp

x = 10
y = 20

swap(x, y)

print(x)
print(y)
````

## 10. Swap Using `return`

A clean way to swap values through a function is to return the new values.

````python
def swap(a, b):
    return b, a

x = 10
y = 20

x, y = swap(x, y)

print(x)
print(y)
````

## 11. Swap Using a Mutable List

````python
def swap(numbers):
    numbers[0], numbers[1] = numbers[1], numbers[0]

data = [10, 20]

swap(data)

print(data)
````

## 12. Shallow Copy

`copy.copy()` creates a new outer object, but nested objects may still be shared.

````python
import copy

numbers = [10, 20, 30]

new_numbers = copy.copy(numbers)
new_numbers.append(100)

print("Original:", numbers)
print("Copy:", new_numbers)
````

````python
import copy

numbers = [[10, 20], [30, 40]]

new_numbers = copy.copy(numbers)
new_numbers[0].append(100)

print("Original:", numbers)
print("Copy:", new_numbers)
````

### Shallow copy idea

```text
Outer object → copied
Nested objects → may be shared
```

## 13. Deep Copy

`copy.deepcopy()` creates independent nested objects.

````python
import copy

numbers = [[10, 20], [30, 40]]

new_numbers = copy.deepcopy(numbers)
new_numbers[0].append(100)

print("Original:", numbers)
print("Deep Copy:", new_numbers)
````

## 14. Function + Deep Copy

````python
import copy

def add_number(numbers):
    numbers = copy.deepcopy(numbers)
    numbers[0].append(100)

original = [[10, 20], [30, 40]]

add_number(original)

print(original)
````

## 15. Function + Mutable Reference Behavior

````python
def add_number(numbers):
    numbers[0].append(100)

original = [[10, 20], [30, 40]]

add_number(original)

print(original)
````

## 16. Default Arguments

A parameter can have a default value. If the caller doesn't provide it, the default is used.

````python
def greet(name="Guest"):
    print("Hello", name)

greet()
greet("Prajjal")
````

````python
def power(number, exponent=2):
    return number ** exponent

print(power(5))
print(power(5, 3))
````

# Final Cheat Sheet

### Function

```python
def function_name(parameters):
    # process
    return result
```

### `print()` vs `return`

```text
print  → display
return → send a value back
```

### Immutable

```text
int, float, str, tuple
```

Rebinding a parameter inside the function does not change the outside variable.

### Mutable

```text
list, dict, set
```

In-place modifications can be visible outside the function.

### Shallow vs Deep Copy

```text
copy.copy()
→ new outer object
→ nested objects may be shared

copy.deepcopy()
→ independent nested objects
```

### Most important concept

> Python passes object references by sharing. Whether the outside value appears to change depends on whether the object is mutable and whether the function mutates it or simply rebinds its local parameter.
